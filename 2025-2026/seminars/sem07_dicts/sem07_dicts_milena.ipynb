Множества $-$ это достаточно удобная коллекция, чтобы хранить неповторяющиеся значения, однако иногда бывают случаи, когда хранить нужно не просто значение, а маппинг `key` -> `value`.

> **Cловарь (или ассоциативный массив)** $-$ это неупорядоченная структура данных, позволяющая идентифицировать ее элементы не по числовому индексу, а по произвольному.

### *Как создать пустой словарь?*

1. через `{}` (не путаем с множеством!)
2. через `dict()`

````python
dictionary_one = {}
dictionary_two = dict()
````

````python
print(type(dictionary_one), type(dictionary_two))
````

````python
print(dictionary_one, dictionary_two)
````

Общее устройство словаря:
```python
{ключ_1: значение_1, ключ_2: значение_2, ..., ключ_n: значение_n}
```

**Повторим, что такое hashable?**

В Python объект является **hashable**, если:
1. У него есть метод `__hash__()` который возвращает хеш-значение
2. Это хеш-значение **не меняется** в течение всего времени жизни объекта

Хеш-значение — это целое число, которое используется для быстрого сравнения объектов и организации данных в хеш-таблицах (как раз для словарей и множеств).

**Почему ключи словаря должны быть hashable?**

Словари в Python работают на основе хеш-таблиц. Когда вы используете ключ для поиска значения, Python:
1. Вычисляет хеш от ключа
2. Использует этот хеш для быстрого нахождения "корзины" (bucket) где хранится значение
3. Сравнивает ключи в этой корзине для точного совпадения

Если бы ключ мог изменяться (как список), его хеш тоже изменился бы, и словарь не смог бы найти соответствующее значение!

**Какие типы данных являются hashable:**

**Hashable (неизменяемые типы):**
- **`int`**, **`float`**, **`complex`** — числовые типы
- **`str`** — строки
- **`tuple`** — кортежи (только если все элементы внутри тоже hashable!)
- **`frozenset`** — неизменяемое множество
- **`bool`** — булевы значения (True/False)
- **`NoneType`** — None
- **`bytes`** — байтовые строки

**Не-hashable (изменяемые типы):**
- **`list`** — списки
- **`dict`** — словари  
- **`set`** — множества
- **`bytearray`** — изменяемые байтовые массивы

**Важное замечание про кортежи:**
Кортеж сам по себе hashable, но только если **все его элементы** тоже hashable. Если внутри кортежа есть изменяемый объект (list, dict, set), то весь кортеж становится не-hashable.

````python
d = {([1, 2, 3]): "values"}
````

````python
d = {(1, 2, 3): "values"}
````

### *Как добавить элемент в существующий словарь?*

Как в список: `dictionary[key] = item`

````python
print(dictionary_three)

dictionary_three['e'] = 12.70

print(dictionary_three)
````

Так же можно его и поменять: `dictionary[key] = new_item`

**NB!** все ключи в словаре уникальны, поэтому, если вы не хотите менять ключ, когда он уже существует, проверяйте!

````python
new_key = input()
new_item = float(input())

if new_key not in dictionary_three:
    dictionary_three[new_key] = new_item
````

````python
print(dictionary_three)
````

Еще есть `.update`: позволяет обновлять словарь, добавив новые key-value пары.

````python
dictionary_three.update({'f': 2.23, 'g': 2.02})

print(dictionary_three)
````

````python
dictionary_three = {**dictionary_three, **{'f': 2.23, 'g': 2.02}}

print(dictionary_three)
````

````python
params = {"sep": ", ", "end": "!"}  # а вот для чего эти ** еще нужны
words = ["one", "two", "three"]
print(*words, **params)  # советую запомнить, такая конструкция очень часто встречается
````

**`*` (одна звездочка) - распаковка позиционных аргументов**

`*words` распаковывает список (или кортеж) в **отдельные позиционные аргументы**:

```python
words = ["one", "two", "three"]
print(*words)  # эквивалентно print("one", "two", "three")
```

**Как это работает:**
- `*` берет каждый элемент списка и передает его как отдельный аргумент
- Вместо одного аргумента-списка получаем три отдельных строковых аргумента
- Работает со списками, кортежами, генераторами - любыми итерируемыми объектами

**`**` (две звездочки) - распаковка именованных аргументов**

`**params` распаковывает словарь в **именованные аргументы**:

```python
params = {"sep": ", ", "end": "!"}
print(**params)  # эквивалентно print(sep=", ", end="!")
```

**Как это работает:**
- `**` берет каждую пару ключ-значение из словаря
- Ключ становится именем параметра, значение - значением параметра
- Работает только со словарями (или объектами, поддерживающими items())

**Важные правила:**

1. **Порядок важен:** сначала идут позиционные аргументы, потом именованные

2. **Именованные аргументы не могут повторяться:**
   ```python
   print(*words, sep=", ", **params)  # Ошибка! sep уже есть в params
   ```

3. **Ключи словаря должны соответствовать именам параметров функции:**
   ```python
   params = {"separator": ", "}  # Нет такого параметра у print()
   print(*words, **params)  # TypeError: 'separator' is an invalid keyword argument
   ```

### *Как сортировать словарь*

1. по ключу

````python
rooms = {"Pink": "#403", "Space": "#201", "Quail": "#500", "Lime": "#503"}

print(dict(sorted(rooms.items())))
````

2. по значению

````python
from operator import itemgetter

print(dict(sorted(rooms.items(), key=itemgetter(1))))
````

А еще проще через `defaultdict`. По сути, это словарь, у которого на все ключи задан `setdefault`.

````python
from collections import defaultdict

list_of_colors = ['blue', 'red', 'blue', 'green', 'yellow', 'pink', 'black', 'black', 'red', 'black', 'blue']
dict_of_colors_easy = defaultdict(int) # можно передать тип данных, по которому берется дефолтное значение
# для int это 0, для str -- пустая строка, для dict это {} и тд
````

А еще есть `Counter`!

````python
from collections import Counter


words = "what a life what a night what a beautiful beautiful ride".split()
print(words)

counter = Counter(words)
print(counter)
````

Возвращает список кортежей самых часто встречаемых элементов и их значений. Ели таких несколько, то приоритет у тех, которые встретились раньше:

````python
print(counter.most_common(1))
````

````python
print(counter.most_common(2))
````

````python
for word in counter.elements():  # выводятся слова с повторениями в порядке, в котором впервые встретились
    print(word)
````

````python
words = "but i crumble completely when you cry".split()
another_words = "when you look at me like that my darling what did you expect".split()

counter = Counter(words)
another_counter = Counter(another_words)

counter.subtract(another_counter)  # можно по-честному вычесть из одного каунтера другой

print(counter)
````
