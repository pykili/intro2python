## Списки / `list`

Типы данных в питоне делятся на два типа:
*   неизменяемые – `int`, `float`, `str`, `bool` (с ними мы уже знакомы) + `tuple` (с ним знакомимся сегодня)
*   изменяемые – `list` (с ним знакомимся сегодня) и `dict`, `set` (с ними познакомимся в течение модуля)



**Список – это упорядоченная изменяемая коллекция объектов любых типов.** При помощи него мы храним **набор** данных (числа, строки и т.д.) в **определенном** порядке. Данные мы можем менять – добавлять или удалять.



Зачем нужен список? Хранить **набор** данных (числа, строки и т.д.) в **определенном** порядке – с этим одна или две переменные, зарезервированные под одно значение, не справятся.

### Индексация и `len()`

У каждого элемента в списке есть свой **индекс** – номер, под которым элемент стоит в наборе.

**Индексация в питоне с нуля** – «первый» элемент имеет индекс 0.

К каждому элементу можно обратиться по его номеру при помощи квадратных скобок.

````python
list_of_wishes = ['iPhone', 'MacBook', 'Свитшот']

print(list_of_wishes[1])
````

Узнать число элементов в списке можно при помощи функции `len()`.

````python
print(len(list_of_wishes))
````

Какой будет индекс у последнего элемента, если индексация с нуля?

````python
print(list_of_wishes[len(list_of_wishes) - 1])
````

Покороче это можно записать при помощи **отрицательных** чисел – у последнего элемента будет индекс `-1`.

````python
print(list_of_wishes[-1])
````

Значение элемента списка по индексу можно поменять.

````python
list_of_wishes[-1] = 'Отпуск на море'
print(list_of_wishes)
````

### Как перебрать элементы списка? Цикл `for`

А что если нам нужно добавить много значений в список? Или перебрать все их по индексам? Или вывести каждое значение на отдельной строке?

Все это можно сделать при помощи цикла – мы будем использовать теперь цикл `for`.

**Синтаксис цикла `for`**

```python
for Переменная in Список значений:
    # четыре пробела
    Действие/я, которые нужно выполнить, ПОКА переменная в списке значений
```

Что важно:
1. Переменную в цикле **не** нужно инициализировать (создавать) заранее.
2. Устоявшееся значение для переменной – `i`; если оно занято, то `j`, в крайнем случае `k`
3. Двоеточие и пробелы **обязательны**!
4. Пробелов обязательно 4 (четыре)
5. Пробелов перед двоеточием **не** надо!

**Диапазон `range(start, stop, step)`**

Команда `range()` генерирует последовательность чисел. `range()` может принимать от одного до трех аргументов.
- `range(stop)` - если range принимает один аргумент, будут сгенерированы числа от 0 до stop-1.  
- `range(start, stop)` - если range принимает два аргумента, будут сгенерированы числа от start до stop-1.  
- `range(start, stop, step)` - если range принимает три аргумента, будут сгенерированы числа от start до stop-1 с шагом step.

___________

В общем, `range(a, b)` - это `range[a, b)`

Это нам поможет перебрать индексы всех элементов списка – числа от `0` до `len(lst)-1` с шагом `1`.

````python
numbers = [1, 5, 3, 8, 2, -1, 0]

# выводим все числа в списке и их индексы
for i in range(len(numbers)):
    print(i, numbers[i])
````

````python
# альтернативный перебор – сами значения, а не индексы
for i in numbers:
    print(i)
````

````python
numbers = []

# добавляем в список числа от 0 до 10
for i in range(11):
    numbers.append(i)

print(*numbers)
````

### Новый способ вводить данные: метод `split()`

Вообще-то это метод не списков, а строк – но они очень тесно связаны. Метод `split()` по заданному разделителю (в данном случае – пробел) сделает из строки список.

````python
digits = "1 2 3 4 5".split()
print(digits)
````

````python
digits = "1,2,3,4,5".split(",")
print(digits)
````

Если в практике данные вводятся в строку – мы используем именно `split()`.

````python
digits = input().split()
print(digits)
````

А если вводится список чисел и нам нужно работать с данными как с числами, то по идее нам нужно к **каждому** элементу списка применить `int()` – по аналогии с `int(input())`.

Это можно сделать короче – вот так:

````python
digits = list(map(int, input().split()))
print(digits)
````

## Кортежи / `tuple`

**Кортеж – это упорядоченная неизменяемая коллекция объектов любых типов.** По сути – это просто неизменяемая версия списка. Он пригождается в более сложных структурах данных и, например, при кастомных функциях сортировок (об этом мы узнаем позднее).

### Как создать кортеж?

````python
# пустой кортеж
t1 = tuple()
t2 = ()
print(t1, t2)
````

````python
# готовый кортеж
t3 = tuple([1, 2, 3])
print(t3)
````

````python
# из одного элемента
t4 = ('Hello!', )
print(t4)
````

### Что можно делать с кортежем?

1. Индексироваться

````python
print(t3[0])
````

````python
# при попытке изменить элемент будет так
t3[0] = -1
````

2. Узнать его размер

````python
print(len(t4))
````

3. Посчитать число вхождений элемента `x` в кортеже

````python
t3.count(3)
````

## **Дополнительная информация**

Списки можно создавать при помощи **генераторов**.

```python
[{элемент или действие с ним} for {элемент} in {последовательность}]
```


````python
lst = [2**i for i in range(0, 10)]
print(lst)
````

Списки можно создавать при помощи сложения других списков.


````python
sweet = ['конфета', 'шоколадка']
salty = ['крекер', 'сухарик']
basket = sweet + salty
print(basket)
````

Что значит **неизменяемый** тип данных? При попытке изменить объект мы либо получим ошибку, либо создадим новый объект с новым адресом в памяти.

````python
a = 4
id1 = id(a)
a = a + 1
id2 = id(a)

# id – это функция, которая позволяет узнать адрес в памяти
print(id1, id2, id1 == id2)
````

Еще немного про не_изменяемость: `list` vs. `tuple`

````python
a_list = [1, 2, 3] # список
a_tuple = (1, 2, 3) # кортеж

print(id(a_list))
print(id(a_tuple))

id_list_before = id(a_list)
id_tuple_before = id(a_tuple)

a_list.extend([3, 4])
a_tuple = a_tuple + (3, 4)
id_list_after = id(a_list)
id_tuple_after = id(a_tuple)

print(id_list_after == id_list_before)
print(id_tuple_after == id_tuple_before)
````

Если вам нужно скопировать данные из одного списка в другой, будьте осторожны – вот так копировать плохо!

````python
lst1 = [1, 2, 3]
lst2 = lst1

lst1[0] = 0

print(lst1, id(lst1))  # ячейки памяти одинаковые
print(lst2, id(lst2))
````

Копию лучше делать вот так:

````python
list_six = [1, 2, 3]
list_seven = list_six.copy()
list_six[0] = 0

print(list_six, id(list_six))  # ячейки памяти разные
print(list_seven, id(list_seven))
````

Или даже так:

````python
from copy import deepcopy

list_six = [1, 2, 3]
list_seven = deepcopy(list_six)  # самый надежный способ для вложенных списков

list_six[0] = 0

print(list_six, id(list_six))  # ячейки памяти разные
print(list_seven, id(list_seven))
````
