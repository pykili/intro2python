## **Немного о последовательностях**

## **Цикл `while`**

**Циклы** — это конструкции, которые позволяют многократно выполнять один и тот же блок кода.

Цикл `while` работает, пока выполняется условие.

Он состоит из следующих частей:

1. **Ключевое слово `while`**  
2. **Условие** — логическое выражение, которое проверяется перед каждой итерацией цикла.
3. **Тело цикла** — блок кода с отступом, выполняемый, пока условие выполняется (истинно).

**Итерация** — это одно выполнение тела цикла во время выполнения программы.

```python
while Условие/я выполнения тела цикла:
    # четыре пробела
    Дейстие/я, которые нужно выполнить, ПОКА утверждение/я истинно/ы
```

Этот цикл удобен, если переменная в условии меняется в теле цикла. Пример, хочу возводить вводимые числа в квадрат, пока мне не введут 0:

````python
# пример пока не введут 0

n = int(input())

while n != 0:
    print("Принято")
    n = int(input())
````

Цикл может быть **бесконечным**. В таком случае прекратить его выполнение может оператор **break** (ну или ваша ОС).

````python
# пример с поиском гласных

word = input() # apple
vowels = 'aeiou'
i = 0 # счетчик для букв

while i != len(word):
    symbol = word[i]
    if symbol in vowels:
        print(symbol)
    i += 1
````

````python
# пример с вводом пароля
true_password = 'abc'

while True:
    password = input()
    if true_password == password:
        print('Все в порядке! Заходите!')
        break
    else:
        print("Пароль неверный")
````

## **Цикл `for`**

Еще есть цикл `for`. Он нужен, чтобы просматривать элементы коллекции. Его синтаксис такой:
```python
for Переменная in Список значений:
    # четыре пробела
    Дейстие/я, которые нужно выполнить, ПОКА переменная в списке значений
```

Что важно:
1. переменную в цикле **не** нужно инициализировать (создавать) заранее.
2. дефолтное значение для переменной `i`. если оно занято, то `j`, в крайнем случае `k`
3. двоеточие и пробелы **обязательны**!
4. пробелов обязательно 4 (четыре)
5. пробелов перед двоеточием **не** надо!

````python
# ищем гласные с for

word = input()
vowels = 'aeiou'

for i in range(len(word)):
    if word[i] in vowels:
        print(word[i])
````

### **Диапазон `range(start, stop, step)`**

Команда `range()` генерирует последовательность чисел. `range()` может принимать от одного до трех аргументов.

`range(stop)` - если range принимает один аргумент, будут сгенерированы числа от 0 до stop-1.  
`range(start, stop)` - если range принимает два аргумента, будут сгенерированы числа от start до stop-1.  
`range(start, stop, step)` - если range принимает три аргумента, будут сгенерированы числа от start до stop-1 с шагом step.

___________

В общем, `range(a, b)` - это `range[a, b)`

### **`enumerate`**

Синтаксис `enumerate`:
```python
for Переменная №1, Переменная №2 in enumerate(Список значений):
```
Это счетчик, где **переменная №1** $-$ это номер элемента в списке, а **переменная №2** $-$ это сам элемент в списке.

Пример:

````python
english = 'abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ'
````

````python
# пронумеруем буквы английского алфавита

number = 1
for letter in english:
    print(number, letter, sep=' - ')
    number += 1
````

Циклы можно вкладывать друг в друга:

````python
n = 10

for i in range(1, n + 1):
    s = 0
    for j in range(1, i + 1):
        s += j
    print("Sum of numbers from 1 to", i, "is", s)
````
