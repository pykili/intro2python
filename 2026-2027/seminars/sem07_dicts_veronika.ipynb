Множества $-$ это достаточно удобная коллекция, чтобы хранить **неповторяющиеся** значения, однако иногда бывают случаи, когда хранить нужно не просто значение, а пару (ключ, значение) или (`key`, `value`).


**Cловарь** – это неупорядоченная изменяемая структура данных, позволяющая идентифицировать ее элементы не по числовому индексу, а по произвольному.

Есть прямая аналогия из жизни: для каждого слова в словаре (например, английского языка) хранится его определение / значение.

**Ключи** в словаре:

*   уникальные (т.е. в словаре не может быть несколько одинаковых ключей с разными значениями)
*   неизменяемые объекты (т.е. числа, строки, кортежи)

**Значения** в словаре:

*   у разных ключей могут совпадать
*   могут быть любого типа


### **Как создать словарь?**

Пустой словарь создается через пустые фигурные скобки (не путать с множеством!) или через `dict()`.

````python
d1 = {}
d2 = dict()

print(d1, type(d1))
print(d2, type(d2))
````

Создать словарь с данными можно **несколькими** способами.

Вручную через перечисление `{key: value}` в словаре:

````python
d = {'a': 8.17, 'b': 1.49, 'c': 2.78}
print(d)
````

Можно использовать список кортежей через `dict([(key, item)])`:

````python
d = dict([('a', 8.17), ('b', 1.49), ('c', 2.78)])
print(d)
````

Можно использовать функцию `.fromkeys([keys], values)` для того чтобы создать словарь из списка ключей с одним и тем же значением (дефолтно – `None`).

````python
d = dict.fromkeys(['a', 'b', 'c']) # не указали дефолтное значение
print(d)
````

````python
d = dict.fromkeys(['a', 'b', 'c'], 0) # указали дефолтное значение
print(d)
````

Через генератор словарей `{key: item for ... in ...}` (по сути почти то же, что и обычный цикл `for`).

````python
d = {a: str(a) + '-й раз' for a in range(1, 7)}
print(d)
````

### Как добавлять элементы в существующий словарь?

Через синтаксис `dictionary[key] = value`.

````python
d = {}
d[0] = "zero"
print(d)
````

Аналогично, значение по ключу меняется через такой же синтаксис: `dictionary[key] = new_value`.

**NB!** все ключи в словаре уникальны, поэтому, если вы не хотите менять ключ, когда он уже существует, проверяйте!

````python
d = {}
d[0] = "zero"

new_key = int(input())
new_value = input()

if new_key not in d:
    d[new_key] = new_value
else:
    print("такой ключ уже есть!")
````

Еще есть метод `.update()`: он позволяет обновлять словарь, добавив новые key-value пары из другого словаря.

````python
d = dict([('a', 8.17), ('b', 1.49), ('c', 2.78)])

d.update({'f': 2.23, 'g': 2.02})

print(d)
````

### Как напечатать элементы словаря?

Если нам нужно печатать конкретное значение по известному ключу, то просто через синтаксис `d[key]`.

````python
d = dict([('a', 8.17), ('b', 1.49), ('c', 2.78)])

print(d['a'])
````

Если какого-то ключа нет, то будет ошибка (проверяйте оператором `in`).

````python
d = dict([('a', 8.17), ('b', 1.49), ('c', 2.78)])

print(d['f'])
````

Либо можно использовать метод `.get()` – он возвращает соответствующее значение или, если нет ключа, `None` (можно задать свое дефолтное).

````python
d = dict([('a', 8.17), ('b', 1.49), ('c', 2.78)])

print(d.get('f'))
````

Все значения словаря можно перебрать через цикл `for` (при помощи «прохода по коллекции», ведь индексов тут, как в списке нет).

````python
d = dict([('a', 8.17), ('b', 1.49), ('c', 2.78)])

for key in d:
    print(key, d[key])
````

Мы можем сразу переписать пары «ключ-значение» через `dictionary.items()`.

````python
d = dict([('a', 8.17), ('b', 1.49), ('c', 2.78)])

print(d.items())
````

````python
d = dict([('a', 8.17), ('b', 1.49), ('c', 2.78)])

for key, value in d.items():
    print(f'Буква {key} встречается в английских текстах в {value}% случаев')
````

Отдельно можно получить список всех ключей

````python
d = dict([('a', 8.17), ('b', 1.49), ('c', 2.78)])

print(d.keys())
````

или список всех значений.

````python
d = dict([('a', 8.17), ('b', 1.49), ('c', 2.78)])

print(d.values())
````

## Другие методы и особенности словарей

Базово можно узнать **размер словаря** (число пар) через `len()`.

````python
d = dict([('a', 8.17), ('b', 1.49), ('c', 2.78)])

print(len(d))
````

**Удаление ключа** производится через `dictionary.pop(key, value)`. Аналогично `.get()`, только с удалением ключа. Если ключа нет, все так же вернется то, что записано вторым аргументом (или дефолтное `None`).

````python
d = dict([('a', 8.17), ('b', 1.49), ('c', 2.78)])

print(d.pop('e', -1))
print(d.pop('a', -1))
````

Можно задавать **дефолтное значение** через `dict.setdefault(key[, default])`. Он возвращает значение по ключу, если ключ существует, либо дефолтное значение (по умолчанию `None`). Если ключ отсутствует, то в словарь добавляется пара `key-default`.

````python
d = {'red': 1}

print(d.setdefault('red', 2))
print(d)

print(d.setdefault('orange', 2))
print(d)

print(d.setdefault('black'))
print(d)
````

Еще есть некоторая особенность, связанная со сравнением словарей. Со множествами это, кстати, работает так же: сравнение происходит без учета порядка.

````python
d = {"name": "Veronika", "surname": "Tsareva"}

print(d == {"surname": "Tsareva", "name": "Veronika"})
````

### Частотные словари: `defaultdict` и `Counter`

Если мы хотим считать какую-то статистику по корпусу, то для удобства можно использовать `defaultdict`. По сути, это словарь, у которого на все ключи задан `setdefault`.

````python
from collections import defaultdict

list_of_colors = ['blue', 'red', 'blue', 'green', 'yellow', 'pink', 'black', 'black', 'red', 'black', 'blue']

dict_of_colors_easy = defaultdict(int) # можно передать тип данных, по которому берется дефолтное значение
# для int это 0, для str - пустая строка, для dict это {} и т.д.

for color in list_of_colors:
    dict_of_colors_easy[color] += 1

print(dict_of_colors_easy)
````

Дополнительно есть `Counter` – сам считает статистику и возвращает список кортежей самых часто встречаемых элементов и их значений. Ели таких несколько, то приоритет у тех, которые встретились раньше.

````python
from collections import Counter


words = "what a life what a night what a beautiful beautiful ride".split()
print(words)

counter = Counter(words)
print(counter)

print(counter.most_common(2))

for word in counter.elements():  # выводятся слова с повторениями в порядке, в котором впервые встретились
    print(word)
````

### «Распаковка» словаря

**`*` (одна звездочка) - распаковка позиционных аргументов**

`*words` распаковывает список (или кортеж) в **отдельные позиционные аргументы**:

```python
words = ["one", "two", "three"]
print(*words)  # эквивалентно print("one", "two", "three")
```

**Как это работает:**
- `*` берет каждый элемент списка и передает его как отдельный аргумент
- Вместо одного аргумента-списка получаем три отдельных строковых аргумента
- Работает со списками, кортежами, генераторами - любыми итерируемыми объектами

**`**` (две звездочки) - распаковка именованных аргументов**

`**params` распаковывает словарь в **именованные аргументы**:

```python
params = {"sep": ", ", "end": "!"}
print(**params)  # эквивалентно print(sep=", ", end="!")
```

**Как это работает:**
- `**` берет каждую пару ключ-значение из словаря
- Ключ становится именем параметра, значение – значением параметра
- Работает только со словарями (или объектами, поддерживающими `items()`)

**Важные правила:**

1. **Порядок важен:** сначала идут позиционные аргументы, потом именованные

2. **Именованные аргументы не могут повторяться:**
   ```python
   print(*words, sep=", ", **params)  # Ошибка! sep уже есть в params
   ```

3. **Ключи словаря должны соответствовать именам параметров функции:**
   ```python
   params = {"separator": ", "}  # Нет такого параметра у print()
   print(*words, **params)  # TypeError: 'separator' is an invalid keyword argument
   ```

### Напоминание про хеш-функции

**Повторим, что такое hashable?**

В Python объект является **hashable**, если:
1. У него есть метод `__hash__()` который возвращает хеш-значение
2. Это хеш-значение **не меняется** в течение всего времени жизни объекта

Хеш-значение — это целое число, которое используется для быстрого сравнения объектов и организации данных в хеш-таблицах (как раз для словарей и множеств).

**Почему ключи словаря должны быть hashable?**

Словари в Python работают на основе хеш-таблиц. Когда вы используете ключ для поиска значения, Python:
1. Вычисляет хеш от ключа
2. Использует этот хеш для быстрого нахождения «корзины» (bucket), где хранится значение
3. Сравнивает ключи в этой корзине для точного совпадения

Если бы ключ мог изменяться (как список), его хеш тоже изменился бы, и словарь не смог бы найти соответствующее значение!

**Какие типы данных являются hashable:**

**Hashable (неизменяемые типы):**
- **`int`**, **`float`**, **`complex`** — числовые типы
- **`str`** — строки
- **`tuple`** — кортежи (только если все элементы внутри тоже hashable!)
- **`frozenset`** — неизменяемое множество
- **`bool`** — булевы значения (True/False)
- **`NoneType`** — None
- **`bytes`** — байтовые строки

**Не-hashable (изменяемые типы):**
- **`list`** — списки
- **`dict`** — словари  
- **`set`** — множества
- **`bytearray`** — изменяемые байтовые массивы

**Важное замечание про кортежи:**
Кортеж сам по себе hashable, но только если **все его элементы** тоже hashable. Если внутри кортежа есть изменяемый объект (`list`, `dict`, `set`), то весь кортеж становится не-hashable.
