### **Форматирование и f-строки**

1. Есть такой особый вид строк $-$ форматированные строки (formatted strings) или f-строки (f-strings). Они требуют букву `f` перед открывающей кавычкой, и позволяют удобнее работать с выводом. Так, внутрь фигурных скобок `{...}` можно передавать переменные.

Подробнее читайте [тут](https://docs.python.org/3/tutorial/inputoutput.html#fancier-output-formatting).

#### **Спецификаторы формата в f-строках**

**Синтаксис**

```python
f'{выражение:формат}'
```

###### `f` - Числа с плавающей точкой (float)
```python
pi = 3.141592653589793
print(f'{pi:.3f}')  # Вывод: 3.142
```
- Округляет до указанного количества знаков после запятой
- `.3f` = 3 знака после запятой

`e` - Экспоненциальная запись

```python
number = 1234.5678
print(f'{number:.2e}')  # Вывод: 1.23e+03
```



Разделители тысяч  

```python
print(f'{1234567:,d}')       # "1,234,567"
```

Знаки

```python
print(f'{number:+.2f}')      # "+1234.57"
```

Выравнивание

```python
print(f'{number:<10.2f}')    # "1234.57   " (влево)

print(f'{number:>10.2f}')    # "   1234.57" (вправо)

print(f'{number:^10.2f}')    # " 1234.57  " (по центру)
```

````python
number = 10
````

````python
print(f'{number:<10.2f}')
````

### **Условные операторы**

Программы, которые мы видели раньше, имели линейную структуру: все действия выполняются обязательно и в той последовательности, что мы написали их в скрипте.

Иногда нам хочется получать разное поведение в зависимости от какого-либо условия.

Например, мы хотим написать программу, которая продает билет в кино только людям, которым есть 16. Для таких случаев в Python есть условный оператор **`if`**.

Выглядит он следующим образом:

```python
if условие:
    инструкция_1
```

Если мы хотим не просто ничего не делать, в случае, когда покупателю меньше 16, а хотя бы написать предупреждение, мы можем воспользоваться оператором **`if... else`**.

А если мы хотим дополнительно обработать ситуацию когда покупателю от 16 до 18: например, предупредить о жестоких сценах в фильме - можно воспользоваться оператором **`if... elif... else`**

В таком случае условия в `if` и `elif` проверяются последовательно.

Выполняется только блок, соответствующий **первому** из истинных условий. Если все проверяемые условия ложны, то выполняется блок `else` (если он присутствует).

Это называется **каскадными условиями**.

Блоки `else` и `elif` при этом необязательные.

#### **Вложенные условия**

Внутри условных операторов можно использовать любые блоки кода, например, все те же условные операторы, чтобы получить двойную вложенность.

Например, представьте, что вы пытаетесь проверить является ли год високосным.



> **Как определить, что год високосный:**   
• Номер года делится на 4 без остатка, но при этом не делится на 100.  
• Номер года делится на 400 без остатка (например, 2000 год).

Ниже словами опишем структуру (да, не самой оптимальной) программы, которая определяет, является ли год високосным:

1. Если год делится на 4 без остатка, перейдите на шаг 2. В противном случае перейдите к выполнению действия 5.
2. Если год делится на 100 без остатка, перейдите на шаг 3. В противном случае перейдите к выполнению действия 4.
3. Если год делится на 400 без остатка, перейдите на шаг 4. В противном случае перейдите к выполнению действия 5.
4. Год високосный (366 дней).
5. Год не високосный год (365 дней).

И, собственно, реализуем её:

````python
year = int(input())

if year % 4 == 0:  # можно еще просто if not year % 4:
    if year % 100 == 0:
        if year % 400 == 0:
            print('Год високосный')
        else:
            print('Год не високосный')
    else:
        print('Год високосный')
else:
    print('Год не високосный')
````

##### **Немного о моржах**

**Моржовый оператор** - оператор, который совмещает операцию **присваивания** и **возврата** значения переменной.

Давайте упростим конструкцию ниже за счет возможностей Python 3.8:

````python
tutor_name = input()

if tutor_name == "Ксения":
    print("Я в группе 1.")
else:
    print(f"Кажется {tutor_name} - семинарист другой группы")
````

Вот так:

````python
if (tutor_name := input()) == "Ксения":
    print("Я в группе 1.")
else:
    print(f"Кажется {tutor_name} - семинарист другой группы")
````

Но вообще моржовый оператор очень коварен и может вводить в заблуждение. Вокруг него в целом было много шумихи в сообществе языка, и это спровоцировало уход из комитета самого Гвидо Ван Россума.

````python
if tutor_name := input() == "Ксения":  # что станет если мы забудем про скобки?
    print("Я в группе 1.")
else:
    print(f"Кажется {tutor_name} - семинарист другой группы")
````
